# Keputusan kami dibiarkan saja karena datanya sudah lengkap dan tidak ada kesalahan di data tipenya

```python
k = df[["PM2.5", "Noise_dB", "Temperature_C"]]

# 1. Tabel Utama (Dapat 7 Metrik)
display(k.describe())

# 2. Pelengkap (4 Metrik Sisanya)
print("=== MODUS ===")
print(k.mode().iloc[0])

print("\n=== VARIANS ===")
print(k.var())

print("\n=== JANGKAUAN (RANGE) ===")
print(k.max() - k.min())

print("\n=== IQR (Q3 - Q1) ===")
print(k.quantile(0.75) - k.quantile(0.25))
```

```python
# 1. Hitung jumlah frekuensi dan persentase
frekuensi = df["City"].value_counts()
persentase = df["City"].value_counts(normalize=True) * 100

# 2. Gabungkan jadi satu tabel rapi
tabel_kategorik = pd.DataFrame({
    'Jumlah (Frekuensi)': frekuensi,
    'Persentase (%)': persentase.round(2)
})

display(tabel_kategorik)

# 3. Cetak modusnya
print("Kategori paling sering muncul (Modus):", df["City"].mode()[0])
```

```python
import matplotlib.pyplot as plt

plt.hist(df["Traffic_density"], bins=20)
plt.title("Sebaran Kepadatan Kendaraan")
plt.xlabel("Jumlah Kendaraan per km²")
plt.ylabel("Frekuensi")
plt.show()
```

Bentuk sebarannya adalah miring ke kanan karena kebanyakan data tersebar di sekitar 80–170 kendaraan/km², sementara ekor kurva memanjang ke kanan sampai sekitar 300 kendaraan/km².

Interpretasi:
- Sebagian besar wilayah mempunyai kepadatan kendaraan sekitar 80–170 kendaraan per km².
- Terdapat sebagian kecil wilayah yang memiliki kepadatan kendaraan di atas 200 kendaraan per km².

```python
plt.hist(df["Noise_dB"], bins=20)

plt.title("Sebaran Tingkat Kebisingan")
plt.ylabel("Frekuensi")
plt.xlabel("Noise (dB)")
plt.show()
```

Bentuk sebarannya adalah simetris. Data tersebar dengan seimbang di dua sisi.

Interpretasi:
- Sebagian besar tingkat kebisingan berada di sekitar 55–75 dB, dengan frekuensi tertinggi adalah 65–70 dB.
- Hanya sebagian kecil kasus tingkat kebisingan sangat rendah (di bawah 40 dB) atau sangat tinggi (di atas 90 dB).

```python
rata = df.groupby("City")["Traffic_density"].mean().reset_index()
plt.bar(rata["City"], rata["Traffic_density"], width=0.5)

plt.title('Tingkat Kepadatan Kendaraan per Kota')
plt.xlabel('Kota')
plt.ylabel('Jumlah Kendaraan per km2')
plt.xticks(rotation=20, ha='right')
plt.show()
```

- Sebaran tingkat kepadatan kendaraan tertinggi berada pada kota Berlin dengan jumlah kendaraan sedikit di atas 120 kendaraan per km².   
- Sebaran tingkat kepadatan kendaraan di kota-kota besar cukup seimbang, yakni sekitar 118-122 kendaraan per km².

```python
plt.boxplot(df["Noise_dB"])
plt.title("Tingkat Kebisingan")
plt.ylabel("Noise (dB)")
plt.show()
```

Kebanyakan data tingkat kebisingan berada di antara tingkat 58 dB hingga 72 dB, dengan nilai median berupa 65.2 dB (median diambil dari data 50% dari df.describe() pada nomor 6).

Untuk menentukan adanya pencilan data atau tidak, hitung batas atas dan batas bawah:

Kuartil 1 (Q1) = 58 dB dan Kuartil 3 (Q3) = 72 dB (diambil dari data 25% dan 75% dari df.describe() pada nomor 6 dan dibulatkan).   

Rentang antarkuartil (IQR) = Q3 - Q1 = 72 - 58 = 14 dB.

Batas bawah = Q1 - (1,5 * IQR) = 58 - (1,5 * 14) = 37 dB. 
Batas atas = Q3 + (1,5 * IQR) = 72 + (1,5 * 14) = 93 dB. 

Data-data yang berada di bawah 37 dB dan di atas 93 dB berada di luar batas bawah maupun atas menurut aturan 1.5 * IQR, sehingga data-data ini merupakan pencilan.
